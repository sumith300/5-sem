# NLP Lab 6: Character-Level Dataset Processing
## Tiny Shakespeare Dataset

This notebook covers character-level tokenization and batching:
1. Load dataset from HuggingFace
2. Character-level tokenization using UTF-8 split
3. Build vocabulary from all characters
4. Create (current_char, next_char) pairs
5. Batch data for training

```python
import datasets
```

```python
# Load dataset
d = datasets.load_dataset('tiny_shakespeare')['train']
print("✓ Dataset loaded from HuggingFace")
```

## Step 1: Character-Level Tokenization

```python
d = d.map(lambda x: datasets.Value('string').unicode_split(x['text'], 'UTF-8'))
print("✓ Character-level tokenization (UTF-8 split) complete")
```

## Step 2: Build Vocabulary

```python
# train split includes vocabulary for other splits
vocabulary = sorted(set(next(iter(d)).numpy()))
print(f"✓ Vocabulary built: {len(vocabulary)} unique characters")
```

## Step 3: Create (current_char, next_char) Pairs

```python
d = d.map(lambda x: {'cur_char': x[:-1], 'next_char': x[1:]})
d = d.unbatch()
print("✓ Created (current_char, next_char) pairs and unbatched")
```

## Step 4: Batch Data for Training

```python
seq_len = 100
batch_size = 2
d = d.batch(seq_len)
d = d.batch(batch_size)
print(f"✓ Batched: seq_len={seq_len}, batch_size={batch_size}")
```

## Summary

```python
print("="*60)
print("DATASET READY FOR TRAINING")
print("="*60)
print(f"✓ Dataset loaded from HuggingFace (karpathy/tiny_shakespeare)")
print(f"✓ Character-level tokenization (UTF-8 split)")
print(f"✓ Vocabulary size: {len(vocabulary)} characters")
print(f"✓ Created (current_char, next_char) pairs")
print(f"✓ Batched with seq_len={seq_len}, batch_size={batch_size}")
print("="*60)
```
